```python
import pandas as pd
```

```python
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
df
```

```python
cols = df[["Name", "Sex", "Age", "Fare", "Survived"]]
print(cols)
```

```python
filtered_df = df[
    (df["Sex"] == "female") & (df["Fare"] > 30)
]

print(filtered_df)
```

```python
grouped_df = df.groupby("Pclass").agg(
    survival_rate = ("Survived", "mean"),
    average_fare = ("Fare", "mean"),
    average_age = ("Age", "mean")
)

grouped_df["survival_rate"] *= 100
print(grouped_df)
```
