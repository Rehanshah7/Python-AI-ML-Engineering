```python
import requests
from bs4 import BeautifulSoup
```

```python
# Fetch data and Store it

page_count = 1;

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("no valid pages anymore...")
        break

    with open(f"scrapedData/quotes{page_count}.html", "w", encoding="utf-8") as f:
        f.write(res.text)
        print(f"Downloaded content from page{page_count}")
    
    page_count = page_count + 1
```

```python
# Extract useful info and find the life quote

page_count = 1
life_quote = []

while True:
    try:
        with open(f"scrapedData/quotes{page_count}.html", "r") as f:
            html_content = f.read()

    except FileNotFoundError:
        print(f"No such file found. Stopped at page {page_count - 1}.")
        break
    
    soup = BeautifulSoup(html_content, "lxml")
    all_quotes = soup.select("div.quote")

    if not all_quotes:
        break

    for q in all_quotes:
        all_tags = []
    
        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text(strip=True))
    
        if "life" in all_tags:
            text = q.select_one("span.text").get_text(strip=True)
            author = q.select_one("small.author").get_text(strip=True)
            life_quote.append([text, author])

    page_count += 1

print("Total life quotes:", len(life_quote))
print(life_quote)
```
