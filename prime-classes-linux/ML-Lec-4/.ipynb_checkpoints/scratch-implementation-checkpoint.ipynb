```python
import numpy as np
```

```python
# Linear Regression
class LinearRegression:
    def __init__(self, learning_rate=0.01, number_iter=1000):
        self.bias = None
        self.weights = None
        self.learning_rate = learning_rate
        self.number_iter = number_iter

    def fit(self, X, y): #X_train, y_train
        m, n = X.shape

        #Step - 1
        self.bias = 0
        self.weights = np.zeros(n)

        #Gradient Descent
        for i in range(self.number_iter):
            #Step - 2
            y_pred = self.bias + np.dot(X, self.weights)
    
            #Step - 3
            derivative_with_respect_to_bias = (1/m) * np.sum(y_pred - y)
            derivative_with_respect_to_weight = (1/m) * np.dot(X.T, (y_pred - y))
    
            #Step - 4
            self.bias = self.bias - self.learning_rate * derivative_with_respect_to_bias
            self.weights = self.weights - self.learning_rate * derivative_with_respect_to_weight
    
    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)

print(model.bias)
print(model.weights)
```
