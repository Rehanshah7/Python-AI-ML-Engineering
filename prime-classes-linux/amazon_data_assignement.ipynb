```python
import json
```

```python
# Load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)

    return data
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
# Clean & structure the data
def clean_data(data):
    text_to_num = {
    "zero": 0, "one": 1, "two": 2, "three": 3, "four": 4,
    "five": 5, "six": 6, "seven": 7, "eight": 8, "nine": 9
    }

    cleaned_data = []
    unique_users = set()
    
    for user in data:
        # Clean ratings
        raw_rating = str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

        # Missing data
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        # Deduplicate Data
        if(user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
```

```python
cleaned_data = clean_data(data)
print(cleaned_data)
```

```python
# Get meaningful insights
def get_insights(data):
    total = 0
    for user in data:
        total += float(user["rating"])

    print(f"Average of all ratings = {total / len(data)}")

    poor_ratings = 0
    for user in data:
        if(float(user["rating"]) < 3):
            poor_ratings += 1

    print(f"Percentage of users in poor ratings: {poor_ratings / len(data) * 100}%")
```

```python
get_insights(cleaned_data)
```

```python
# Recommondation feature for the product
def get_recommondations(data):
    recommondations = []
    for user in data:
        curr_recommondations = {}
        curr_recommondations["name"] = user["name"]
        if(float(user["rating"]) >= 4):
            curr_recommondations["brand"] = "Apple"
        else:
            curr_recommondations["brand"] = "Samsung"

        recommondations.append(curr_recommondations)
    return recommondations
```

```python
recommonded_data = get_recommondations(cleaned_data)
print(recommonded_data)
```
